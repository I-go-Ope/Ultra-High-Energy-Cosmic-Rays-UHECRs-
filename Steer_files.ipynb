```python
import numpy as np
import os

codefile = './Downloads/particlecodes.csv'
rawcodes = np.loadtxt(codefile, delimiter=",", dtype=str, skiprows=1)

# columns: species number, species number, CORSIKA particle code
codes = np.vstack((rawcodes[:, 0], rawcodes[:, 0], rawcodes[:, 3])).transpose()

rawcodes
```

```python
nshower = 100

base_outdir = os.path.expanduser("~/Downloads/Asteer")

# CORSIKA energies (GeV)
energies = {
    15: "1.E6",
    16: "1.E7",
    17: "1.E8",
    18: "1.E9",
    19: "1.E10",
    20: "1.E11",
    21: "1.E12",
}

for energy_level, energy in energies.items():

    energy_folder = os.path.join(base_outdir, f"E{energy_level}")
    os.makedirs(energy_folder, exist_ok=True)

    for code in codes:

        species_number = int(code[0])
        corsika_code = code[2]

        runnr = species_number

        seed1 = species_number
        seed2 = species_number + 100
        seed3 = species_number + 200

        filename = f"E{energy_level}.{species_number}.steer"
        filepath = os.path.join(energy_folder, filename)

        outstr = f"""RUNNR   {runnr}
EVTNR   1
NSHOW   {nshower}

PRMPAR  {corsika_code}
ERANGE  {energy} {energy}

THETAP  20. 20.
PHIP    -180. 180.

SEED    {seed1} 0 0
SEED    {seed2} 0 0
SEED    {seed3} 0 0

OBSLEV  110.E2

MAGNET  20.0 42.8

EPOPAR input ../epos/epos.param
EPOPAR fname pathnx ../epos/
EPOPAR fname inics ../epos/epos.inics
EPOPAR fname iniev ../epos/epos.iniev
EPOPAR fname initl ../epos/epos.initl
EPOPAR fname inirj ../epos/epos.inirj
EPOPAR fname hpf   ../epos/urqmd34/tables.dat
EPOPAR fname check none
EPOPAR fname histo none
EPOPAR fname data  none
EPOPAR fname copy  none

HADFLG  0 0 0 0 0 2

ECUTS   0.3 0.3 0.003 0.003

ELMFLG  T T

LONGI   T 10. T T

THIN    1.E-6 1.E2 0.

MAXPRT  1

CERARY  1 1 1. 1. 100. 100.
CWAVLG  300 600
CERFIL  1

DIRECT  ./

EXIT
"""

        with open(filepath, "w") as f:
            f.write(outstr)

        print(f"Created {filepath}")
```
